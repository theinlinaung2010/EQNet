```python
import torch
import torch.nn as nn
import torch.nn.functional as F


class ImageClassifierModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.conv1 = nn.Conv2d(1, 1, kernel_size=3, padding=1)

    def forward(self, x: torch.Tensor):
        x = self.conv1(x)
        x = F.relu(x)
        return x
```

```python
torch_model = ImageClassifierModel()
# Create example input for exporting the model
example_input = torch.randn(1, 1, 32, 16)
onnx_program = torch.onnx.export(
    torch_model, 
    example_input, 
    dynamo=True,
    input_names=['input'],
    output_names=['output']
)
```

```python
onnx_program.save("sample_CNN_model.onnx")
```
