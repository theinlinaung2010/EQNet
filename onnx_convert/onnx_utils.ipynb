```python
import onnx_utils as ou
import onnx
import datetime
```

```python
model_path = r"./phasenet_das_test.onnx"
export_path = model_path.replace(".onnx", "_export.onnx")

names_to_replace = [
    ("data", "input_strain_rate"),
    ("phase", "output_phase")
]

metadata = {
    "name": "PhaseNet-DAS ONNX Model (for real-time inference)",
    "description": "PhaseNet-DAS-v1 model is converted from PyTorch to ONNX format for real-time inference on DAS data.",
    "inference_classes": ["Noise", "P", "S"],
    "base_model": "PhaseNet-DAS-v1",    
    "input_data_type": "Strain Rate",
    "inference_window_sec": 10.24,
    "interrogation_rate_sps": 100,
    "author": "Aung",
    "datetime": datetime.datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
}

model = onnx.load(model_path)
model = ou.rename_onnx_node(model, names_to_replace)
model = ou.add_metadata(model, metadata)

onnx.save(model, export_path)
```
